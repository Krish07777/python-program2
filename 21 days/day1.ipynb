### Reverse a Number

```python
def reverse_number(n):
    rev = 0 
    while n > 0:
        rev = rev * 10 + n % 10
        n //=10
    return rev

print(reverse_number(156))
```

### Power of a Number (xⁿ)

```python
def power(x, n):
    if n == 0:
        return 1
    return x * power(x, n-1)

print(power(2,4))
```

# 16-02-2026

### Matrix Addition

```python
def matrix_add(a,b):
    return [[a[i][j] + b[i][j] for j in range(len(a[0]))] for i in range(len(a))]

print(matrix_add([[1,2],[3,4]], [[5,6],[7,8]]))
```

### Perfect Number

```python
num = int(input("entre number: "))
sum_div = 0

for i in range(1, num):
    if num % i ==0:
        sum_div += i

if sum_div == num:
    print("perfect number")
else:
    print("not perfect number")
```

### Find First Non-Repeating Character

```python
text = input("entre string: ")

freq = {}

for char in text:
    freq[char] = freq.get(char, 0)+1

for char in text:
    if freq[char] == 1:
        print("frist non repeating character:", char)
        break

else: 
    print("no unique found")
```
